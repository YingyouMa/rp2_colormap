# Optimal color immersion for 3D nematic directors

## 1. From an ideal coloring to an optimal immersion

A director field in a 3D nematic liquid crystal is represented by a **unit director**

$$
\mathbf{n}=(n_x,n_y,n_z),\qquad\mathbf{n}\cdot\mathbf{n}=1,
$$

with the nematic identification

$$\mathbf{n}\sim-\mathbf{n}.$$

Thus the physical orientation space is

$$\mathbb{S}^2/\{\mathbf{n}\sim-\mathbf{n}\}\cong\mathbb{RP}^2.$$

For visualization, we would like a color assignment that is nematically consistent, continuous, and globally one-to-one. This would amount to an embedding

$$\mathbf{c}:\mathbb{RP}^2\hookrightarrow\mathbb{R}^3.$$

However,

$$\boxed{\mathbb{RP}^2\not\hookrightarrow\mathbb{R}^3}.$$

We therefore relax global uniqueness and instead seek an **immersion**: global self-intersections are allowed, while nearby orientations should not be locally collapsed. A classical example is Boy's surface.

> **Among realizable immersions of the nematic orientation space into 3D color space, find one that is optimal for scientific visualization.**


## 2. What does an optimal immersion mean?

The final map must lie inside the displayable sRGB gamut,

$$
\mathbf{c}_{\mathrm{sRGB}}(\mathbf{n})=(R,G,B),\qquad 0\le R,G,B\le1.
$$

Within this admissible set, several desirable properties compete, so the problem is naturally multi-objective.

### 2.1 Metric fidelity

For two physical director orientations, let $d_Q$ be their physical distance and $d_c$ the perceptual distance between their assigned colors. We would like

$$d_c(\mathbf n,\mathbf m)^2\approx\alpha d_Q(\mathbf n,\mathbf m)^2,$$

with the same $\alpha>0$ throughout orientation space. We therefore define

$$
\mathcal J_{\mathrm{pair}}=\iint\left[d_c(\mathbf n,\mathbf m)^2-\alpha d_Q(\mathbf n,\mathbf m)^2\right]^2\,d\mu(\mathbf n)d\mu(\mathbf m).
$$

We choose

$$d_Q(\mathbf n,\mathbf m)^2=1-(\mathbf n\cdot\mathbf m)^2,$$

for which

$$1-(\mathbf n\cdot\mathbf m)^2=\frac23\left[1-P_2(\mathbf n\cdot\mathbf m)\right].$$

Thus, apart from the constant term, the orientation dependence lies in the $\ell=2$ sector. For color distance we work in OKLab,

$$\mathbf c=(L,a,b),\qquad d_c(\mathbf n,\mathbf m)=\|\mathbf c(\mathbf n)-\mathbf c(\mathbf m)\|.$$

### 2.2 Lightness uniformity

We prefer small

$$\mathcal J_L=\operatorname{Var}_{\mathbf n}[L(\mathbf n)].$$

Exact constant lightness is impossible for a genuine immersion because it would confine the image to a single $L=L_0$ plane, effectively reducing the target to $\mathbb R^2$, into which $\mathbb{RP}^2$ cannot be immersed.

### 2.3 Reference directions and intuitive colors

We also introduce target colors $\mathbf c_R,\mathbf c_G,\mathbf c_B$ for the three Cartesian director axes,

$$
\mathcal J_{\mathrm{RGB}}=\|\mathbf c(\mathbf e_x)-\mathbf c_R\|^2+\|\mathbf c(\mathbf e_y)-\mathbf c_G\|^2+\|\mathbf c(\mathbf e_z)-\mathbf c_B\|^2.
$$

The axis-to-color assignment is a visual convention, and the three targets need not be the sRGB primaries. Their numerical values will be chosen later.

The intended application is visualization on a white background, so colors should not approach white too closely.

## 3. Strategy of this project

The optimization is organized in two stages.

### Stage I: optimize a prescribed Boy-type immersion

We begin with a known polynomial immersion $\mathbf p_B$ and optimize its affine placement in perceptual color space,

$$\mathbf c_B([\mathbf n])=\mathbf A\,\mathbf p_B(\mathbf n)+\boldsymbol\beta.$$

Provided that $\mathbf A$ remains nonsingular, this preserves the immersion property. Stage I therefore isolates the question of how well a prescribed Boy-type geometry can be tuned as a colormap.

### Stage II: optimize over a broader family of maps

Stage II searches over a broader family, for example finite expansions in even spherical harmonics. Such maps are smooth but are not automatically immersions everywhere: their differential can lose rank at particular orientations. The immersion condition must therefore be checked or enforced throughout orientation space.

## 5. Preparation before optimization

Before optimization, we simplify some utility functions and make the parameters appearing in them explicit.

### 5.1 Harmonic form of the pairwise metric-fidelity loss

Define

$$
R(\mathbf n,\mathbf m)=\|\mathbf c(\mathbf n)-\mathbf c(\mathbf m)\|^2-\alpha\left[1-(\mathbf n\cdot\mathbf m)^2\right].
$$

Let $\Phi_{\ell m}$ be real spherical harmonics orthonormal with respect to $d\mu=d\Omega/(4\pi)$. Since $R$ is a function on $\mathbb S^2\times\mathbb S^2$, it can be expanded in the product basis,

$$
R(\mathbf n,\mathbf m)=\sum_{\ell,m}\sum_{\ell',m'}r_{\ell m,\ell' m'}\,\Phi_{\ell m}(\mathbf n)\Phi_{\ell' m'}(\mathbf m).
$$

In general these sums may be infinite. Because the product basis is orthonormal, Parseval's identity gives

$$
\boxed{\mathcal J_{\mathrm{pair}}=\sum_{\ell,m}\sum_{\ell',m'}\left|r_{\ell m,\ell' m'}\right|^2}.
$$

A particularly useful case occurs when $R$ contains only finitely many product-harmonic modes: the series above then terminates exactly. Whether this happens is controlled mainly by the color map $\mathbf c(\mathbf n)$, because the director-distance term is already finite,

$$
1-(\mathbf n\cdot\mathbf m)^2=\frac23\left[1-P_2(\mathbf n\cdot\mathbf m)\right],
$$

so it contains only the two harmonic sectors $\ell=0$ and $\ell=2$. Therefore, if $\mathbf c$ has a finite spherical-harmonic expansion, then $R$ also has a finite expansion and $\mathcal J_{\mathrm{pair}}$ becomes an exact finite sum.

### 5.2 Reference colors with comparable lightness

The standard sRGB primaries have different OKLab lightnesses, so we do not use them directly as the three reference colors.

Instead, we keep the **OKLab hue directions** of the sRGB red, green, and blue primaries, and look for three colors with the same lightness and the same chroma. For a candidate common lightness $L$, let $C_R^{\max}(L)$, $C_G^{\max}(L)$, and $C_B^{\max}(L)$ be the largest chroma values that remain inside the sRGB gamut along the three fixed hue directions. The largest common chroma available at that lightness is

$$
C_{\mathrm{common}}(L)
=
\min\left\{
C_R^{\max}(L),
C_G^{\max}(L),
C_B^{\max}(L)
\right\}.
$$

We then choose

$$
L_* = \operatorname*{arg\,max}_L C_{\mathrm{common}}(L),
\qquad
C_* = C_{\mathrm{common}}(L_*).
$$

Numerically,

$$
L_*\approx0.61305505,
\qquad
C_*\approx0.20860695.
$$

The resulting ideal OKLab anchors are

$$
\tilde{\mathbf c}_R=(0.61305505,\ 0.18203739,\ 0.10187859),
$$

$$
\tilde{\mathbf c}_G=(0.61305505,\ -0.16548869,\ 0.12700533),
$$

$$
\tilde{\mathbf c}_B=(0.61305505,\ -0.02161699,\ -0.20748389).
$$

After conversion to sRGB and rounding to 8-bit display values, these become

$$
(230,60,47),\qquad(0,161,0),\qquad(65,121,255).
$$

Converting those rounded 8-bit colors back to OKLab gives

$$
\mathbf c_R=(0.61194632,\ 0.18219108,\ 0.10119720),
$$

$$
\mathbf c_G=(0.61430529,\ -0.16582618,\ 0.12726434),
$$

$$
\mathbf c_B=(0.61340896,\ -0.02144615,\ -0.20727421).
$$

The small differences in lightness are therefore only the result of 8-bit rounding. The calculation is reproduced below.


```python
import numpy as np
from scipy.optimize import minimize_scalar

# OKLab conversion matrices
M1 = np.array([
    [0.4122214708, 0.5363325363, 0.0514459929],
    [0.2119034982, 0.6806995451, 0.1073969566],
    [0.0883024619, 0.2817188376, 0.6299787005],
])
M2 = np.array([
    [0.2104542553, 0.7936177850, -0.0040720468],
    [1.9779984951, -2.4285922050, 0.4505937099],
    [0.0259040371, 0.7827717662, -0.8086757660],
])
M1_inv = np.linalg.inv(M1)
M2_inv = np.linalg.inv(M2)

def srgb_to_linear(c):
    c = np.asarray(c, dtype=float)
    return np.where(c <= 0.04045, c / 12.92,
                    ((c + 0.055) / 1.055) ** 2.4)

def linear_to_srgb(c):
    c = np.asarray(c, dtype=float)
    return np.where(c <= 0.0031308, 12.92 * c,
                    1.055 * np.power(np.clip(c, 0, None), 1 / 2.4) - 0.055)

def linear_srgb_to_oklab(rgb):
    return M2 @ np.cbrt(M1 @ np.asarray(rgb, dtype=float))

def oklab_to_linear_srgb(lab):
    return M1_inv @ (M2_inv @ np.asarray(lab, dtype=float)) ** 3

# Hue directions of the three sRGB primaries in the OKLab a-b plane
primaries = {
    "R": np.array([1.0, 0.0, 0.0]),
    "G": np.array([0.0, 1.0, 0.0]),
    "B": np.array([0.0, 0.0, 1.0]),
}
hue_dir = {}
for name, rgb in primaries.items():
    lab = linear_srgb_to_oklab(srgb_to_linear(rgb))
    hue_dir[name] = lab[1:] / np.linalg.norm(lab[1:])

def in_gamut(linear_rgb, tol=1e-10):
    return np.all(linear_rgb >= -tol) and np.all(linear_rgb <= 1 + tol)

# Maximum in-gamut chroma at fixed L and fixed hue direction
def max_chroma_at_L(L, direction):
    lo, hi = 0.0, 1.5
    while in_gamut(oklab_to_linear_srgb([L, *(hi * direction)])):
        hi *= 2

    for _ in range(80):
        mid = 0.5 * (lo + hi)
        lab = [L, *(mid * direction)]
        if in_gamut(oklab_to_linear_srgb(lab)):
            lo = mid
        else:
            hi = mid
    return lo

def common_chroma(L):
    return min(max_chroma_at_L(L, hue_dir[k]) for k in "RGB")

# Maximize the common chroma over the common lightness
result = minimize_scalar(
    lambda L: -common_chroma(L),
    bounds=(0.45, 0.80),
    method="bounded",
    options={"xatol": 1e-13},
)

L_star = result.x
C_star = common_chroma(L_star)

print(f"L* = {L_star:.8f}")
print(f"C* = {C_star:.8f}")

for name in "RGB":
    lab = np.array([L_star, *(C_star * hue_dir[name])])
    srgb = linear_to_srgb(oklab_to_linear_srgb(lab))
    rgb8 = np.rint(np.clip(srgb, 0, 1) * 255).astype(int)
    lab8 = linear_srgb_to_oklab(srgb_to_linear(rgb8 / 255))

    print(f"\n{name}")
    print("ideal OKLab =", lab)
    print("sRGB x 255  =", srgb * 255)
    print("8-bit sRGB =", rgb8)
    print("rounded OKLab =", lab8)
```
