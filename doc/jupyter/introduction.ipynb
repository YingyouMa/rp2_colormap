# Optimal color immersion for 3D nematic directors

## 1. From an ideal coloring to an optimal immersion

A director field in a 3D nematic liquid crystal is represented by a **unit director**

$$
\mathbf{n}=(n_x,n_y,n_z),
\qquad
\mathbf{n}\cdot\mathbf{n}=1,
$$

with the nematic identification

$$
\mathbf{n}\sim-\mathbf{n}.
$$

Thus the physical orientation space is not the sphere itself. The space of unit vectors is $\mathbb{S}^2$, but antipodal points represent the same nematic orientation. The physical orientation space is therefore

$$
\mathbb{S}^2/\{\mathbf{n}\sim-\mathbf{n}\}
\cong\mathbb{RP}^2,
$$

the real projective plane.

For visualization, we would like to assign a color to every physical director orientation. At the most idealized level, a satisfactory coloring should satisfy three basic requirements:

- **Nematic consistency.** $\mathbf{n}$ and $-\mathbf{n}$ must be assigned exactly the same color. Mathematically, the coloring must be well defined on $\mathbb{RP}^2$.
- **Uniqueness.** Two distinct physical director orientations should receive two distinct colors. Mathematically, the color map should be one-to-one.
- **Continuity.** A small change in director orientation should produce a small change in color. Mathematically, the color map should be continuous.

A color is ultimately described by three coordinates, so at this level we may regard color space as a subset of $\mathbb{R}^3$. In mathematical language, the requirements above amount to searching for an **embedding**

$$
\mathbf{c}:\mathbb{RP}^2\hookrightarrow\mathbb{R}^3.
$$

Roughly speaking, an embedding places one space inside another without tearing it, introducing discontinuities, or identifying two distinct points. For the coloring problem, it would mean that the entire nematic orientation space could be represented continuously in 3D color space while every distinct physical orientation retained its own unique color.

However, it is mathematically impossible to embed $\mathbb{RP}^2$ in $\mathbb{R}^3$:

$$
\boxed{\mathbb{RP}^2\not\hookrightarrow\mathbb{R}^3}.
$$

This is a fundamental topological obstruction rather than a limitation of a particular colormap construction. Since the physical RGB cube is itself only a subset of $\mathbb{R}^3$, restricting the target to realizable RGB colors cannot remove the obstruction. A continuous, globally unique three-coordinate coloring of all 3D nematic orientations is therefore impossible.

We consequently relax the requirement of global uniqueness. Distinct, sufficiently separated orientations are allowed to map to the same point in color space, but the map should still preserve the local two-dimensional structure of the orientation space. Infinitesimally, two independent changes of director orientation should remain two independent changes in color.

For a smooth map $\mathbf{c}:\mathbb{RP}^2\to\mathbb{R}^3$, this means requiring

$$
\operatorname{rank}\!\left(D\mathbf{c}_{[\mathbf{n}]}\right)=2
\qquad
\forall\,[\mathbf{n}]\in\mathbb{RP}^2.
$$

Such a map is called an **immersion**. Unlike an embedding, an immersion may have global self-intersections, but it does not locally collapse the orientation space into a curve or a point.

Crucially, immersions of $\mathbb{RP}^2$ into $\mathbb{R}^3$ do exist. A classical example is **Boy's surface**, which is a smooth immersion of the real projective plane in 3D with unavoidable self-intersections.

The central problem of this project can therefore be stated as follows:

> **Among realizable immersions of the nematic orientation space into 3D color space, find one that is optimal for scientific visualization.**

The remaining question is what *optimal* should mean perceptually and physically.


## 2. What should an optimal color immersion preserve?

Once global one-to-one correspondence is known to be impossible, the problem becomes an optimization problem rather than a search for an exact representation.

For two unit nematic directors $\mathbf{n}$ and $\mathbf{m}$, a natural sign-invariant physical distance is

$$
d_Q\!\left(\mathbf{n},\mathbf{m}\right)^2=1-\left(\mathbf{n}\cdot\mathbf{m}\right)^2.
$$

This distance vanishes both for $\mathbf{m}=\mathbf{n}$ and for $\mathbf{m}=-\mathbf{n}$, as required by nematic symmetry. It is also directly related to the Frobenius distance between uniaxial $Q$ tensors at fixed scalar order parameter.

Ideally, perceptually similar colors should correspond to physically similar directors, while physically distant directors should receive perceptually distinct colors. We therefore seek a map for which color-space distance approximately follows $d_Q$ over the entire orientation space.

A representative global metric objective is

$$
\mathcal{J}_{\mathrm{pair}}=\iint_{\mathbb{RP}^2\times\mathbb{RP}^2}\left[\left\|\mathbf{c}([\mathbf{n}])-\mathbf{c}([\mathbf{m}])\right\|^2-\alpha\left(1-\left(\mathbf{n}\cdot\mathbf{m}\right)^2\right)\right]^2\,\mathrm{d}\mu([\mathbf{n}])\,\mathrm{d}\mu([\mathbf{m}]),
$$

where $\mathrm{d}\mu$ is the uniform orientational measure and $\alpha$ sets the overall scale relating physical and perceptual distances.

Metric fidelity is not the only requirement. For scientific visualization we also care about:

- approximately uniform perceptual lightness, so that apparent brightness does not create artificial structure;
- intuitive reference colors for important directions, for example assigning the Cartesian axes to colors near red, green, and blue;
- sufficient contrast on a white background;
- confinement to the displayable sRGB gamut;
- preservation of the immersion condition, except for the unavoidable global self-intersections.

These requirements compete with one another. In particular, forcing nearly constant lightness tends to flatten the image of $\mathbb{RP}^2$ toward a two-dimensional plane in color space, whereas metric fidelity may prefer to use all three perceptual dimensions. The relevant object is therefore a controlled trade-off among physically and perceptually meaningful objectives.


## 3. RGB coordinates versus perceptual color space

The final colors must be displayable as sRGB values, but Euclidean distance in raw RGB coordinates is not a good model of perceived color difference. RGB should therefore be understood as the final display representation rather than necessarily the space in which the geometry is optimized.

The working strategy of this project is to construct the immersion in the 3D perceptual color space **OKLab**,

$$\mathbf{c}=(L,a,b),$$

and require the resulting colors to lie inside the sRGB gamut after conversion.

This separates two issues:

1. the geometry of the map from nematic orientation space into perceptual color space;
2. the hardware/display constraint that the resulting colors must be realizable in sRGB.

The intended application is scientific visualization on a white background, so perceptual lightness and contrast against white are treated as explicit design considerations rather than accidental consequences of the map.


## 4. Strategy of this project

The optimization is organized in two stages.

### Stage I: optimize a prescribed Boy-type immersion

We begin with a known analytic immersion of $\mathbb{RP}^2$ related to Boy's surface. Rather than immediately searching over all possible maps, Stage I keeps this immersion geometry fixed and optimizes how it is affinely placed in perceptual color space:

$$\mathbf{c}_B([\mathbf{n}])=\mathbf{A}\,\mathbf{p}_B(\mathbf{n})+\boldsymbol{\beta},$$

where $\mathbf{p}_B$ is a fixed polynomial immersion, $\mathbf{A}\in\mathbb{R}^{3\times3}$ is a linear transformation, and $\boldsymbol{\beta}\in\mathbb{R}^3$ is a translation in color space.

This stage asks:

> How good a nematic colormap can be obtained by optimally tuning a prescribed Boy-type immersion?

### Stage II: optimize over a broader family of immersions

A later stage removes the Boy-surface prior and optimizes over a broader family of smooth maps satisfying nematic symmetry,

$$\mathbf{c}(-\mathbf{n})=\mathbf{c}(\mathbf{n}),$$

for example through an expansion in even spherical harmonics. Comparing the best Stage-I map with the best map from this broader class will help distinguish limitations caused by color tuning from limitations intrinsic to the prescribed immersion geometry.


## 5. Reproducibility

The purpose of this repository is not only to obtain a visually useful colormap, but to make the construction reproducible and auditable. Symbolic reductions, numerical optimization, parameter scans, and validation tests should be generated from the stated mathematical definitions rather than copied from hand-derived intermediate coefficients.

The intended workflow is:

- use exact symbolic algebra where the polynomial structure permits it;
- derive spherical moments from the original map rather than hard-coding derived coefficients;
- convert the resulting symbolic objectives into numerical functions for optimization;
- validate analytic expressions independently by direct numerical sampling;
- record numerical precision, software versions, optimizer settings, tolerances, initial conditions, and random seeds;
- keep generated results reproducible from scripts or notebooks.

The long-term objective is a colormap that represents 3D nematic orientation as perceptually faithfully, interpretably, and displayably as the topology of $\mathbb{RP}^2$ allows.
