# Optimal color immersion for 3D nematic directors

## 1. From an ideal coloring to an optimal immersion

A director field in a 3D nematic liquid crystal is represented by a **unit director**

$$
\mathbf{n}=(n_x,n_y,n_z),\qquad\mathbf{n}\cdot\mathbf{n}=1,
$$

with the nematic identification

$$\mathbf{n}\sim-\mathbf{n}.$$

Thus the physical orientation space is not the sphere itself. The space of unit vectors is $\mathbb{S}^2$, but antipodal points represent the same nematic orientation. The physical orientation space is therefore

$$\mathbb{S}^2/\{\mathbf{n}\sim-\mathbf{n}\}\cong\mathbb{RP}^2,$$

the real projective plane.

For visualization, we would like to assign a color to every physical director orientation. At the most idealized level, a satisfactory coloring should satisfy three basic requirements:

- **Nematic consistency.** $\mathbf{n}$ and $-\mathbf{n}$ must be assigned exactly the same color. Mathematically, the coloring must be well defined on $\mathbb{RP}^2$.
- **Uniqueness.** Two distinct physical director orientations should receive two distinct colors. Mathematically, the color map should be one-to-one.
- **Continuity.** A small change in director orientation should produce a small change in color. Mathematically, the color map should be continuous.

A color is ultimately described by three coordinates, so at this level we may regard color space as a subset of $\mathbb{R}^3$. In mathematical language, the requirements above amount to searching for an **embedding**

$$\mathbf{c}:\mathbb{RP}^2\hookrightarrow\mathbb{R}^3.$$

Roughly speaking, an embedding places one space inside another without tearing it, introducing discontinuities, or identifying two distinct points. For the coloring problem, it would mean that the entire nematic orientation space could be represented continuously in 3D color space while every distinct physical orientation retained its own unique color.

However, it is mathematically impossible to embed $\mathbb{RP}^2$ in $\mathbb{R}^3$:

$$\boxed{\mathbb{RP}^2\not\hookrightarrow\mathbb{R}^3}.$$

This is a fundamental topological obstruction rather than a limitation of a particular colormap construction. Since the physical RGB cube is itself only a subset of $\mathbb{R}^3$, restricting the target to realizable RGB colors cannot remove the obstruction. A continuous, globally unique three-coordinate coloring of all 3D nematic orientations is therefore impossible.

We consequently relax the requirement of global uniqueness. Distinct, sufficiently separated orientations are allowed to map to the same point in color space, while locally distinct orientations should remain locally distinguishable. In mathematical language, this changes the problem from searching for an embedding to searching for an **immersion** of $\mathbb{RP}^2$ into $\mathbb{R}^3$. Roughly speaking, an immersion is allowed to intersect itself globally, but it should not locally collapse a neighborhood of orientations into a lower-dimensional object.

Crucially, such immersions do exist. A classical example is **Boy's surface**, an immersion of the real projective plane in 3D with unavoidable self-intersections.

The central problem of this project can therefore be stated as follows:

> **Among realizable immersions of the nematic orientation space into 3D color space, find one that is optimal for scientific visualization.**


## 2. What does an optimal immersion mean?

Once we allow immersions rather than embeddings, the problem is no longer the existence of a map. There are infinitely many ways to immerse $\mathbb{RP}^2$ in $\mathbb{R}^3$. The question is therefore which immersion gives the most useful representation of nematic orientation as color.

### The display constraint

There is one immediate hard constraint. The final result is intended to be displayed as an ordinary RGB image, so every color produced by the map must be representable in the chosen RGB color space. For sRGB, after conversion to normalized sRGB coordinates,

$$
\mathbf{c}_{\mathrm{sRGB}}(\mathbf{n})=(R,G,B),
\qquad 0\le R,G,B\le1
\qquad\forall\,[\mathbf{n}]\in\mathbb{RP}^2.
$$

In other words, every mapped color must lie inside the displayable sRGB gamut. A color outside this gamut cannot be shown faithfully on an sRGB display.

Within this admissible set, however, there is no single criterion that completely determines the best colormap. Several desirable properties compete with one another. We therefore formulate the problem as a **multi-objective optimization**: each desirable property is represented by a utility, or equivalently a loss, and the aim is to identify **Pareto-optimal** compromises. A solution is Pareto optimal if no utility can be improved without making at least one other utility worse.

The utilities considered in this project are introduced below.

### 2.1 Metric fidelity

The first objective is to make color differences reflect director differences as faithfully as possible. For two physical director orientations $\mathbf{n}$ and $\mathbf{m}$, let

$$d_Q(\mathbf{n},\mathbf{m})$$

denote their physical director distance, and let

$$d_c(\mathbf{n},\mathbf{m})$$

denote the perceptual distance between their assigned colors. At this point we leave the concrete definitions of these two distances open.

If two director orientations are very similar, their colors should look similar. As the director orientations become more different, their colors should also appear progressively more different. Ideally, perceptual color distance should increase proportionally with physical director distance. Since the objective below is formulated in terms of squared distances, we introduce a single positive scale factor $\alpha$ and write

$$
d_c(\mathbf{n},\mathbf{m})^2\approx\alpha\,d_Q(\mathbf{n},\mathbf{m})^2.
$$

The same $\alpha$ is used throughout the entire orientation space. Equally importantly, this relation should be as uniform and isotropic as possible: the same physical director distance should correspond to approximately the same perceptual color distance regardless of where the pair lies in $\mathbb{RP}^2$.

This suggests evaluating the deviation from the ideal proportional relation over **all pairs of orientations**, rather than only over a selected set of directions. Using the uniform orientation measure $\mathrm d\mu$, we define the metric-fidelity loss as

$$
\mathcal{J}_{\mathrm{pair}}
=\iint
\left[
d_c(\mathbf{n},\mathbf{m})^2
-\alpha d_Q(\mathbf{n},\mathbf{m})^2
\right]^2
\,\mathrm d\mu(\mathbf{n})\,\mathrm d\mu(\mathbf{m}),
$$

where $\alpha>0$ sets the overall scale between the two squared metrics. The integration over the full orientation space makes the objective sensitive to spatially nonuniform or anisotropic distortions: a map cannot obtain a small loss merely by representing a few preferred directions well.

The remaining question is therefore concrete: **what should we use for $d_Q$ and $d_c$?** Our choices are as follows.

#### Director distance $d_Q$

For two unit directors $\mathbf{n}$ and $\mathbf{m}$, we choose

$$
d_Q(\mathbf{n},\mathbf{m})^2=1-(\mathbf{n}\cdot\mathbf{m})^2.
$$

A useful reason for this choice is its direct connection to the $\ell=2$ spherical-harmonic sector that describes nematic orientational order. Since

$$
1-(\mathbf{n}\cdot\mathbf{m})^2
=\frac{2}{3}\left[1-P_2(\mathbf{n}\cdot\mathbf{m})\right],
$$

the squared distance is expressed entirely through the second Legendre polynomial. By the spherical-harmonic addition theorem, $P_2(\mathbf{n}\cdot\mathbf{m})$ is a finite sum of products of $\ell=2$ spherical harmonics. This gives the metric a particularly simple harmonic structure. The importance of this choice is not only conceptual: later, when evaluating $\mathcal{J}_{\mathrm{pair}}$ for polynomial or finite spherical-harmonic maps, this finite low-order structure will allow the double integral to be reduced to a finite collection of spherical moments and will greatly simplify the exact calculation.

#### Perceptual color distance $d_c$: OKLab

For the color distance, we need a numerical measure of how different two colors appear to the human eye. Raw RGB coordinates are not suitable for this purpose: RGB values specify the intensities of display primaries, and Euclidean distance in RGB does not reliably correspond to perceived color difference.

We therefore represent the immersion in **OKLab**, a perceptual color space designed so that ordinary Euclidean distance approximately tracks perceived color difference. An OKLab color is written as

$$
\mathbf{c}=(L,a,b),
$$

where $L$ represents perceptual lightness and $a$ and $b$ are two chromatic coordinates. We choose

$$
d_c(\mathbf{n},\mathbf{m})
=\|\mathbf{c}(\mathbf{n})-\mathbf{c}(\mathbf{m})\|
=\sqrt{(\Delta L)^2+(\Delta a)^2+(\Delta b)^2},
$$

as an approximate measure of perceptual color difference. The optimization is therefore carried out in OKLab, while sRGB enters separately through the display-gamut constraint stated above.

With these choices, the abstract metric-fidelity loss becomes

$$
\mathcal{J}_{\mathrm{pair}}
=\iint
\left[
\|\mathbf{c}(\mathbf{n})-\mathbf{c}(\mathbf{m})\|^2
-\alpha\left(1-(\mathbf{n}\cdot\mathbf{m})^2\right)
\right]^2
\,\mathrm d\mu(\mathbf{n})\,\mathrm d\mu(\mathbf{m}).
$$

Smaller $\mathcal{J}_{\mathrm{pair}}$ means better global metric fidelity.

### 2.2 Lightness uniformity

For scientific visualization, changes in apparent brightness can create visual structure that is unrelated to the underlying director field. Since $L$ is the perceptual-lightness coordinate of OKLab, we prefer $L(\mathbf{n})$ to vary as little as possible over the orientation space. A natural loss is

$$
\mathcal{J}_{L}=\operatorname{Var}_{\mathbf{n}}[L(\mathbf{n})].
$$

Smaller $\mathcal{J}_L$ corresponds to more uniform perceptual lightness, but $\mathcal{J}_L=0$ cannot be attained by a genuine immersion. If the variance were exactly zero, then $L(\mathbf{n})=L_0$ would be constant over the entire orientation space, so every mapped color would lie in the single constant-lightness plane

$$
\mathbf{c}(\mathbb{RP}^2)\subset\{(L_0,a,b)\}\cong\mathbb{R}^2.
$$

Thus exact constant lightness would reduce the effective target from 3D color space to a 2D plane. But $\mathbb{RP}^2$ cannot be immersed in $\mathbb{R}^2$. Consequently, exact zero lightness variance is incompatible with the immersion requirement: lightness can be made nearly uniform, but some nonzero variation in $L$ is mathematically necessary.

### 2.3 Reference directions and intuitive colors

A scientifically useful colormap should also be easy to interpret. In particular, it is convenient for the three Cartesian director axes to correspond to familiar reference colors near red, green, and blue. If $\mathbf{c}_R$, $\mathbf{c}_G$, and $\mathbf{c}_B$ denote the chosen target colors expressed in OKLab, a corresponding loss is

$$
\mathcal{J}_{\mathrm{RGB}}=\|\mathbf{c}(\mathbf{e}_x)-\mathbf{c}_R\|^2+\|\mathbf{c}(\mathbf{e}_y)-\mathbf{c}_G\|^2+\|\mathbf{c}(\mathbf{e}_z)-\mathbf{c}_B\|^2.
$$

The assignment of the Cartesian axes to red, green, and blue is only a convenient visual convention, not a mathematical requirement. Other assignments, or other sets of reference colors altogether, could equally well be used, provided that the selected reference colors are sufficiently separated perceptually. Moreover, even when red-, green-, and blue-like colors are chosen, they need not be the three sRGB primaries. The sRGB primaries have substantially different perceptual lightnesses, whereas comparable lightness is desirable for the present colormap. The reference colors $\mathbf{c}_R$, $\mathbf{c}_G$, and $\mathbf{c}_B$ are therefore simply parameters of this utility and may be chosen according to the desired visualization. Their specific numerical values will be discussed later.

### 2.4 Visibility on a white background

The intended application is visualization on a white background. Colors that approach white too closely can become difficult to see even if the immersion itself has good internal metric properties. We therefore also prefer sufficient perceptual contrast between the mapped colors and white.

## 3. Strategy of this project

The optimization is organized in two stages.

### Stage I: optimize a prescribed Boy-type immersion

We begin with a known analytic immersion of $\mathbb{RP}^2$ related to Boy's surface. Rather than immediately searching over all possible maps, Stage I keeps this immersion geometry fixed and optimizes how it is affinely placed in perceptual color space:

$$\mathbf{c}_B([\mathbf{n}])=\mathbf{A}\,\mathbf{p}_B(\mathbf{n})+\boldsymbol{\beta},$$

where $\mathbf{p}_B$ is a fixed polynomial immersion, $\mathbf{A}\in\mathbb{R}^{3\times3}$ is a linear transformation, and $\boldsymbol{\beta}\in\mathbb{R}^3$ is a translation in color space. Provided that $\mathbf{A}$ remains nonsingular, this affine transformation preserves the immersion property. This is the principal reason for beginning with a Boy-type map: $\mathbf{p}_B$ is already known to be an immersion over the entire orientation space, so Stage I can focus on color optimization without having to construct the local immersion property from scratch.

This stage asks:

> How good a nematic colormap can be obtained by optimally tuning a prescribed Boy-type immersion?

### Stage II: optimize over a broader family of maps

Stage II removes the Boy-surface prior and searches over a substantially broader family of smooth maps, for example maps constructed from finite expansions in even spherical harmonics. This gives the optimization much more geometric freedom than an affine transformation of one prescribed Boy-type surface.

However, an important complication appears at this stage. A map built from spherical harmonics is not automatically an immersion everywhere. Even if the map is smooth as a function, there may be particular orientations or local regions where its differential loses rank, so that nearby director orientations are locally collapsed rather than represented by two independent directions in color space. Such a map would fail the immersion requirement even though its global objective values might otherwise look favorable.

This distinction explains the two-stage strategy. In Stage I, the starting Boy-type map is already known to be an immersion globally, and any nonsingular linear transformation of its 3D image preserves that property. In Stage II, by contrast, the immersion condition becomes an additional property that must be verified or enforced throughout the entire orientation space while the map itself is being optimized. The precise mathematical and numerical procedure used to guarantee an everywhere-immersive map will therefore be developed when Stage II is analyzed in detail.

Comparing the best Stage-I map with the best admissible Stage-II map will then help distinguish limitations caused by color tuning from limitations intrinsic to the prescribed Boy-type geometry.


## 4. Reproducibility

The purpose of this repository is not only to obtain a visually useful colormap, but to make the construction reproducible and auditable. Symbolic reductions, numerical optimization, parameter scans, and validation tests should be generated from the stated mathematical definitions rather than copied from hand-derived intermediate coefficients.

The intended workflow is:

- use exact symbolic algebra where the polynomial structure permits it;
- derive spherical moments from the original map rather than hard-coding derived coefficients;
- convert the resulting symbolic objectives into numerical functions for optimization;
- validate analytic expressions independently by direct numerical sampling;
- record numerical precision, software versions, optimizer settings, tolerances, initial conditions, and random seeds;
- keep generated results reproducible from scripts or notebooks.

The long-term objective is a colormap that represents 3D nematic orientation as perceptually faithfully, interpretably, and displayably as the topology of $\mathbb{RP}^2$ allows.


## 5. Preparation: simplifying the utility functions

Before carrying out either optimization stage, it is useful to simplify the utility functions as far as possible analytically. The main reason is computational: the optimizer will evaluate the same objectives many times, so expressions that initially appear to require integration over the full orientation space should, whenever possible, be reduced to finite algebraic combinations of precomputable coefficients. This also makes the numerical implementation easier to verify because the reduced formulas can be compared independently with direct numerical integration.

We begin with the most expensive-looking objective, the pairwise metric-fidelity loss.

### 5.1 Harmonic reduction of the pairwise metric-fidelity loss

Recall

$$
\mathcal{J}_{\mathrm{pair}}
=\iint
\left[
\|\mathbf{c}(\mathbf{n})-\mathbf{c}(\mathbf{m})\|^2
-\alpha\left(1-(\mathbf{n}\cdot\mathbf{m})^2\right)
\right]^2
\,\mathrm d\mu(\mathbf{n})\,\mathrm d\mu(\mathbf{m}).
$$

At first sight this is a 4D integral over two independent orientations. For the finite polynomial and finite spherical-harmonic maps considered in this project, however, the entire integrand has a finite spherical-harmonic representation. The double integral can therefore be reduced exactly to a finite sum.

To make the normalization convenient, let $\Phi_{\ell m}$ denote real spherical harmonics orthonormal with respect to the uniform probability measure on the sphere,

$$
\mathrm d\mu=\frac{\mathrm d\Omega}{4\pi},
\qquad
\int \Phi_{\ell m}(\mathbf{n})\,\Phi_{\ell' m'}(\mathbf{n})\,\mathrm d\mu(\mathbf{n})
=\delta_{\ell\ell'}\delta_{mm'}.
$$

With this convention, the addition theorem becomes

$$
P_{\ell}(\mathbf{n}\cdot\mathbf{m})
=\frac{1}{2\ell+1}
\sum_{m=-\ell}^{\ell}
\Phi_{\ell m}(\mathbf{n})\,\Phi_{\ell m}(\mathbf{m}).
$$

The director-distance term is especially simple because

$$
1-(\mathbf{n}\cdot\mathbf{m})^2
=\frac{2}{3}
\left[1-P_2(\mathbf{n}\cdot\mathbf{m})\right].
$$

Thus its orientation dependence contains only the $\ell=0$ and $\ell=2$ sectors. This is the key simplification gained by our choice of $d_Q$.

Now define the residual kernel

$$
R(\mathbf{n},\mathbf{m})
=\|\mathbf{c}(\mathbf{n})-\mathbf{c}(\mathbf{m})\|^2
-\alpha\left(1-(\mathbf{n}\cdot\mathbf{m})^2\right).
$$

For a finite polynomial or finite spherical-harmonic map $\mathbf{c}$, $R$ can be expanded in the product basis

$$
R(\mathbf{n},\mathbf{m})
=\sum_{\ell,m}\sum_{\ell',m'}
r_{\ell m,\ell' m'}
\Phi_{\ell m}(\mathbf{n})
\Phi_{\ell' m'}(\mathbf{m}),
$$

with only finitely many nonzero coefficients. Because the product basis is orthonormal on the pair space $\mathbb S^2\times\mathbb S^2$, Parseval's identity gives immediately

$$
\boxed{
\mathcal{J}_{\mathrm{pair}}
=\sum_{\ell,m}\sum_{\ell',m'}
\left|r_{\ell m,\ell' m'}\right|^2
}.
$$

Therefore the original double integral does not need to be evaluated numerically during optimization. Once the finite harmonic coefficients of the map are known, the objective reduces to a finite algebraic expression in those coefficients and in $\alpha$. For the Stage-I Boy polynomial, the required harmonic coefficients can be derived exactly from the polynomial map; for Stage II they are directly related to the spherical-harmonic expansion coefficients being optimized.

The next step is to work out this reduction explicitly for the Stage-I Boy map and express $\mathcal{J}_{\mathrm{pair}}$ in terms of the affine parameters.
